```python
# importando tabela
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")
display(tabela)
```

```python
# visualizando banco de dados
tabela = tabela.drop(columns="CustomerID")
display(tabela)
```

```python
# tratamento de dados
display(tabela.info())

tabela = tabela.dropna()

display(tabela.info())
```

```python
# analise inicial - quantos canelamentos?
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True))
```

```python
# analise detalhada
import plotly.express as px

for coluna in tabela.columns:

  grafico = px.histogram(tabela, x= coluna, color= "cancelou", text_auto=True)
  grafico.show()
```

insights

- todos os clientes que possuiam contrato mensal cancelaram o serviço
  - dar desconto dos contratos anual e trimestral

- ligações callcenter > 4 = todos cancelaram
  - se o cliente ligar 3x, ligar alerta

- atraso de pagamento > 20 dias = cliente cancela
  - ligar alerta se cliente atrasar +15 dias

como calcular taxa de cancelamento após resolver esses 3 problemas?

- filtrar base de dados
  - duracao_contrato: != mensal
  - ligacoes_callcenter: <= 4
  - dias_atraso: <= 20

```python
# mensal
condicao = tabela["duracao_contrato"] != "Monthly"
tabela = tabela[condicao]

# ligações callcenter
condicao = tabela["ligacoes_callcenter"] <= 4
tabela = tabela[condicao]

# dias de atraso
condicao = tabela["dias_atraso"] <= 20
tabela = tabela[condicao]

display(tabela["cancelou"].value_counts(normalize=True))
```
